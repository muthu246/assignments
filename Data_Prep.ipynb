```python
import pandas as pd
from sklearn.preprocessing import LabelEncoder, RobustScaler

# Data
data = {
    'Gender': ['Male', 'Female', 'Female', 'Male', 'Female'],
    'City': ['Mumbai', 'Delhi', 'Chennai', 'Mumbai', 'Delhi'],
    'Size': ['Small', 'Large', 'Medium', 'Large', 'Small'],
    'Age': [25, 45, 32, 60, 28],
    'Fare': [15, 300, 85, 450, 20]
}

df = pd.DataFrame(data)

# -----------------------
# CATEGORICAL ENCODING
# -----------------------

# 1. Gender → Label Encoding
le = LabelEncoder()
df['Gender'] = le.fit_transform(df['Gender'])
# Male/Female → 1/0 (order depends on alphabetical sorting)

# 2. City → One-Hot Encoding (drop first to avoid dummy trap)
df = pd.get_dummies(df, columns=['City'], drop_first=True)

# 3. Size → Manual Label Encoding (ordered)
size_mapping = {'Small': 0, 'Medium': 1, 'Large': 2}
df['Size'] = df['Size'].map(size_mapping)

# -----------------------
# FEATURE SCALING
# -----------------------

# Use RobustScaler because Fare has outliers (e.g., 450)
scaler = RobustScaler()

df[['Age', 'Fare']] = scaler.fit_transform(df[['Age', 'Fare']])
# RobustScaler uses median and IQR, so it is not distorted by extreme values

print(df)
```
